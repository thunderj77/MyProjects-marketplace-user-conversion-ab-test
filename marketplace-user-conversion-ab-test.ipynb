## 2. Анализ ключевых метрик продукта

### Задача

Провести верхнеуровневую оценку текущего состояния продукта на основе агрегированных данных. Без глубокой сегментации – только общая картина.

**Важное допущение:** выручка маркетплейса составляет 5% от стоимости каждого заказа (комиссия платформы).

### Что анализируем

Динамика основных метрик за период:
- Выручка (Revenue)
- Активная аудитория (MAU)
- Новые пользователи (New Users)
- Количество заказов (Orders)
- Конверсия активных пользователей в покупателей (CR)
- Средний чек (AOV)
- Стоимость привлечения (CAC)

### Цель

Быстро получить целостную картину, ответить на вопросы:
- Соответствует ли продукт базовым ожиданиям?
- Наблюдается ли рост?
- Есть ли тревожные сигналы?

```python
# ОЧИСТКА И ПОДГОТОВКА ДАННЫХ
# Удаляем технический столбец-индекс из таблицы campaigns 
campaigns = campaigns.drop(columns=['Unnamed: 0'], errors='ignore')

# Рассчитываем реальную выручку маркетплейса: 5% комиссии от суммы заказа
orders['revenue'] = orders['total_price'] * 0.05

# АГРЕГАЦИЯ МЕТРИК ПО МЕСЯЦАМ
# 1. MAU (месячная активная аудитория) и количество событий
monthly_events = (
    events
    .groupby(events['event_month'].dt.to_period('M'))  # группируем по месяцам
    .agg(
        mau=('user_id', 'nunique'),    # уникальные пользователи в месяц
        events=('event_id', 'count')   # общее число событий
    )
)

# 2. Заказы: количество, покупатели, выручка, общая стоимость продаж
monthly_orders = (
    orders
    .groupby(orders['order_month'].dt.to_period('M'))
    .agg(
        orders=('order_id', 'nunique'),      # количество уникальных заказов
        buyers=('user_id', 'nunique'),       # количество уникальных покупателей
        revenue=('revenue', 'sum'),          # выручка маркетплейса
        total_sales=('total_price', 'sum')   # полная стоимость проданных товаров
    )
)

# 3. Новые пользователи по месяцу регистрации (когортам)
monthly_users = (
    users
    .groupby(users['cohort_month'].dt.to_period('M'))
    .agg(new_users=('user_id', 'nunique'))
)

# 4. Маркетинговые затраты по месяцам
monthly_spend = (
    campaigns
    .groupby(campaigns['spend_month'].dt.to_period('M'))
    .agg(spend=('budget', 'sum'))
)

# 5. Объединяем все витрины в единую таблицу
monthly = (
    monthly_events
    .join(monthly_orders, how='left')   # left = сохраняем все месяцы из events
    .join(monthly_users, how='left')
    .join(monthly_spend, how='left')
    .fillna(0)                           # заменяем NaN на 0 (нет заказов/затрат)
)

# РАСЧЁТ ПРОИЗВОДНЫХ МЕТРИК (ЮНИТ-ЭКОНОМИКА)
# AOV (Average Order Value) – средний чек: общая выручка / количество заказов
monthly['aov'] = (monthly['total_sales'] / monthly['orders'].replace(0, np.nan)).round(2)

# CR (Conversion Rate) – конверсия активных пользователей в покупателей
monthly['cr'] = (monthly['buyers'] / monthly['mau']).round(4)

# CAC (Customer Acquisition Cost) – стоимость привлечения одного нового пользователя
monthly['cac'] = (monthly['spend'] / monthly['new_users'].replace(0, np.nan)).round(2)

# ФОРМАТИРОВАНИЕ ДЛЯ ВЫВОДА
monthly = monthly.reset_index() # сбрасываем индекс (месяц становится колонкой)

# Определяем колонку с месяцем (гибкий поиск, но в данном случае это 'event_month')
period_col = next((col for col in monthly.columns if 'month' in col.lower()), monthly.columns[0])

monthly = monthly.rename(columns={period_col: 'month'})
monthly['month'] = monthly['month'].astype(str)    # конвертируем Period в строку для читаемости
monthly = monthly.sort_values('month')

# Выручку округляем до целых – для наглядной презентации
monthly['revenue'] = monthly['revenue'].round(0).astype(int)

# Выводим только нужные колонки
cols_to_show = ['month', 'mau', 'new_users', 'orders', 'buyers', 'revenue', 'aov', 'cr', 'cac']
available_cols = [c for c in cols_to_show if c in monthly.columns]

print("📊 Ежемесячная динамика ключевых метрик:")
display(monthly[available_cols].round(4))
```

```python
# Настройка стиля
plt.rcParams['figure.figsize'] = (12, 14)
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 10})

# Метрики для визуализации
metrics = [
    ('mau', 'MAU', 'Активные пользователи в месяц', '{:,.0f}'),
    ('new_users', 'New Users', 'Новые пользователи', '{:,.0f}'),
    ('orders', 'Orders', 'Количество заказов', '{:,.0f}'),
    ('revenue', 'Revenue', 'Выручка', '{:,.0f}'),
    ('aov', 'AOV', 'Средний чек', '{:,.0f}'),
    ('cr', 'CR', 'Конверсия в заказ (%)', '{:.1f}%'),
    ('cac', 'CAC', 'Стоимость привлечения', '{:,.0f}')
]

# Создаём subplot 4×2
fig, axes = plt.subplots(4, 2, figsize=(14, 16))
axes = axes.flatten()

# Строим каждый график
for i, (col, title, ylabel, fmt) in enumerate(metrics):
    if col in monthly.columns:
        ax = axes[i]
        # Для CR — умножаем на 100 для отображения в %
        y = monthly[col] * (100 if col == 'cr' else 1)
        
        # Линейный график
        ax.plot(monthly['month'], y, marker='o', linewidth=2, markersize=5, 
                color='#2E86AB', markerfacecolor='white', markeredgewidth=1.5)
        
        # Адаптивные границы
        y_min, y_max = y.min(), y.max()
        y_range = y_max - y_min if y_max != y_min else y_max * 0.1
        ax.set_ylim(y_min - y_range * 0.1, y_max + y_range * 0.15)
        
        # Заливка под графиком
        ax.fill_between(range(len(y)), y, alpha=0.08, color='#2E86AB')
        
        # Подписываем точки
        for j, v in enumerate(y):
            ax.text(j, v + y_range * 0.02, fmt.format(v), 
                   ha='center', va='bottom', fontsize=8)
        
        ax.set_title(title, fontsize=12, fontweight='bold')
        ax.set_ylabel(ylabel)
        ax.set_xlabel('Месяц')
        ax.tick_params(axis='x', rotation=45)
        
        # Убираем верхнюю и правую границы
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

# Удаляем пустой 8-й график
axes[7].set_visible(False)

plt.tight_layout()
plt.suptitle('Динамика ключевых метрик (2024)', fontsize=16, fontweight='bold', y=1.02)
plt.show()
```

**Показатели эффективности маркетинга:**
- CAC (стоимость привлечения одного пользователя)
- ROI (рентабельность инвестиций)
- Payback Period (срок окупаемости)

```python
# Создаём независимую копию датафрейма, чтобы не изменить исходные данные при добавлении новых колонок
monthly_plot = monthly.copy()

# Рассчитываем LTV (Lifetime Value) как среднюю выручку на одного нового пользователя
# Если в месяце не было новых пользователей, ставим NaN — избегаем деления на ноль
monthly_plot['ltv'] = np.where(
    monthly_plot['new_users'] > 0,
    monthly_plot['revenue'] / monthly_plot['new_users'],
    np.nan
)

# Рассчитываем ROI (Return on Investment) в процентах
# Формула: (LTV / CAC - 1) * 100
# Если CAC равен нулю или отрицательный, ставим NaN — расчёт невозможен
monthly_plot['roi'] = np.where(
    monthly_plot['cac'] > 0,
    (monthly_plot['ltv'] / monthly_plot['cac'] - 1) * 100,
    np.nan
)

# Подготавливаем данные для оси X: числовые индексы и подписи месяцев
x = np.arange(len(monthly_plot))       # [0, 1, 2, ...] для позиционирования точек
months = monthly_plot['month'].values  # названия месяцев для подписей
```

```python
# Создаём фигуру с двумя графиками рядом
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# ГРАФИК 1: LTV и CAC
# Строим линии LTV и CAC с маркерами в каждой точке
ax1.plot(x, monthly_plot['ltv'], marker='o', label='LTV', linewidth=2)
ax1.plot(x, monthly_plot['cac'], marker='o', label='CAC', linewidth=2)

# Подписываем значения LTV только в ключевых точках, чтобы не загромождать график
# Выбираем: первый месяц, последний, максимум и минимум (если данных > 4)
highlight_indices = [0, len(x) - 1]  # первый и последний месяц
if len(x) > 4:
    highlight_indices.append(monthly_plot['ltv'].idxmax())  # месяц с макс. LTV
    highlight_indices.append(monthly_plot['ltv'].idxmin())  # месяц с мин. LTV

# Выводим подписи для выбранных точек
for i in highlight_indices:
    # Проверяем, что значение не NaN (например, если в месяце не было пользователей)
    if not np.isnan(monthly_plot['ltv'].iloc[i]):
        # Размещаем текст чуть выше точки (смещение +5 по вертикали)
        ax1.text(i, monthly_plot['ltv'].iloc[i] + 5,
                 f"{monthly_plot['ltv'].iloc[i]:.0f}",  # целое число без знаков после запятой
                 ha='center', va='bottom', fontsize=8)

# Настройка оформления первого графика
ax1.set_title('LTV и CAC по месяцам', fontsize=13, fontweight='bold')
ax1.set_xlabel('Месяц')
ax1.set_ylabel('Значение, ₽')           
ax1.set_xticks(x)                       
ax1.set_xticklabels(months, rotation=45, ha='right') 
ax1.grid(True, alpha=0.3)       
ax1.legend()

# ГРАФИК 2: ROI
# Строим линию ROI с маркерами
ax2.plot(x, monthly_plot['roi'], marker='o', linewidth=2, label='ROI')

# Подписываем значения ROI в каждой точке (их обычно немного, график не перегружен)
for i in x:
    roi_val = monthly_plot['roi'].iloc[i]
    # Пропускаем NaN-значения (месяцы без данных)
    if not np.isnan(roi_val):
        # Формат +XXX.X% — знак плюс/минус и один знак после запятой
        ax2.text(i, roi_val + 2, f"{roi_val:+.1f}%",
                 ha='center', va='bottom', fontsize=8)

# Настройка оформления второго графика
ax2.set_title('ROI по месяцам', fontsize=13, fontweight='bold')
ax2.set_xlabel('Месяц')
ax2.set_ylabel('ROI, %')
ax2.set_xticks(x)
ax2.set_xticklabels(months, rotation=45, ha='right')
ax2.axhline(0, color='gray', linewidth=0.8, linestyle='--')  # линия безубыточности (ROI = 0%)
ax2.grid(True, alpha=0.3)
ax2.legend()

# Автоматически выравниваем отступы между элементами
plt.tight_layout()
plt.show()
```

```python
# 1. Расчёт агрегатов
total_revenue = monthly['revenue'].sum()
total_spend = monthly['spend'].sum()
total_new_users = monthly['new_users'].sum()
total_orders = monthly['orders'].sum()

avg_arpu = (monthly['revenue'] / monthly['mau']).mean()
avg_aov = monthly['aov'].mean()
avg_cac = monthly['cac'].mean()
avg_cr = monthly['cr'].mean()
avg_arppu = avg_aov * 0.05  # 5% комиссии

avg_orders_per_user = total_orders / users['user_id'].nunique()
ltv = avg_orders_per_user * avg_arppu

roi = (ltv - avg_cac) / avg_cac if avg_cac > 0 else np.nan
# Payback
# Предположим, заказы распределены равномерно за 12 месяцев
monthly_rev_per_user = (avg_orders_per_user / 12) * avg_arppu
payback_months = avg_cac / monthly_rev_per_user if monthly_rev_per_user > 0 else np.nan

# 2. Создаём датафрейм
summary = pd.DataFrame({
    'Метрика': [
        'Общая выручка',
        'Общие затраты на привлечение',
        'Новые пользователи за год',
        'Средний чек (AOV)',
        'Выручка на активного пользователя (ARPU)',
        'Выручка на заказ (ARPPU)',
        'Стоимость привлечения (CAC)',
        'Конверсия в заказ (CR)',
        'Пожизненная ценность (LTV)',
        'ROI (LTV/CAC - 1)',
        'Payback Period, мес'
    ],
    'Значение': [
        total_revenue,
        total_spend,
        total_new_users,
        avg_aov,
        avg_arpu,
        avg_arppu,
        avg_cac,
        avg_cr,
        ltv,
        roi,
        payback_months
    ]
})

# 3. Форматируем значения
def format_value(row):
    val = row['Значение']
    metric = row['Метрика']
    if 'выручка' in metric.lower() or 'затраты' in metric.lower() or 'AOV' in metric or 'ARPU' in metric or 'ARPPU' in metric or 'CAC' in metric or 'LTV' in metric:
        return f"{val:,.0f}" if abs(val) >= 1000 else f"{val:.1f}"
    elif 'CR' in metric or 'ROI' in metric:
        return f"{val:.1%}"
    elif 'Payback' in metric:
        return f"{val:.1f}"
    else:
        return str(val)

summary['Значение'] = summary.apply(format_value, axis=1)

print("Итоговая таблица монетизации и юнит-экономики 2024")
display(summary)
```

**Зоны роста.** Удлинение payback и снижение конверсии во второй половине года – ключевые риски, требующие фокуса при масштабировании.


```python
# 1. ПОДГОТОВКА ДАННЫХ ДЛЯ КОГОРТНОГО АНАЛИЗА
# Преобразуем месяц когорты в строку формата 'ГГГГ-ММ' для удобства группировки
users['cohort_month'] = pd.to_datetime(users['cohort_month']).dt.strftime('%Y-%m')

# Соединяем заказы с информацией о пользователе, чтобы узнать месяц его регистрации (когорту)
# Это позволяет отнести каждый заказ к конкретной когорте пользователей
orders_with_cohort = orders.merge(users[['user_id', 'cohort_month']], on='user_id', how='inner')

# Преобразуем даты в datetime
orders_with_cohort['order_date'] = pd.to_datetime(orders_with_cohort['order_date'])
# Извлекаем месяц заказа в формате Period ('ГГГГ-ММ')
orders_with_cohort['order_month'] = orders_with_cohort['order_date'].dt.to_period('M')
# Приводим месяц когорты к такому же формату Period
orders_with_cohort['cohort_month'] = pd.to_datetime(orders_with_cohort['cohort_month']).dt.to_period('M')

# Расчет количества месяцев между заказом и регистрацией
orders_with_cohort['n_month'] = (
    (orders_with_cohort['order_month'] - orders_with_cohort['cohort_month'])
    .apply(lambda x: x.n if not pd.isna(x) else -1) # x.n извлекает числовое значение разницы в месяцах
)

# Оставляем только заказы, сделанные в месяце регистрации или позже (исключаем ошибки)
orders_with_cohort = orders_with_cohort[orders_with_cohort['n_month'] >= 0]

# 2. АГРЕГАЦИЯ ДАННЫХ ДЛЯ ПОСТРОЕНИЯ КОГОРТНОЙ ТАБЛИЦЫ
# Создаем когортную таблицу выручки.
cohort_revenue = (
    orders_with_cohort
    .groupby(['cohort_month', 'n_month'])['revenue'].sum()
    .unstack(fill_value=0) # fill_value=0 заполняет нулями месяцы, в которых не было заказов
)
cohort_revenue.index = cohort_revenue.index.astype(str) # Приводим индекс к строковому типу

# --- Подготовка вспомогательных данных для расчета LTV и ROI ---
# Размер каждой когорты (сколько пользователей было привлечено в этом месяце)
cohort_size = users['cohort_month'].value_counts().sort_index()
# Месяцы из основного датафрейма метрик (приводим к формату 'ГГГГ-ММ')
monthly['month'] = monthly['month'].str[:7]
# CAC (стоимость привлечения) по месяцам, для использования в расчете ROI
monthly_cac = monthly.set_index('month')['cac']

# 3. РАСЧЕТ ROI ДЛЯ КАЖДОЙ КОГОРТЫ И КАЖДОГО МЕСЯЦА ЕЕ ЖИЗНИ
# Находим общие месяцы, для которых есть данные во всех трех источниках
common = cohort_revenue.index.intersection(cohort_size.index).intersection(monthly_cac.index)

# Оставляем только общие данные, чтобы избежать ошибок
cohort_revenue = cohort_revenue.loc[common]
cohort_size = cohort_size.loc[common]
monthly_cac = monthly_cac.loc[common]

# Расчет LTV (Lifetime Value)
cumulative_rev = cohort_revenue.cumsum(axis=1)

# LTV = Накопленная выручка / Размер когорты.
# Делим построчно (axis=0), чтобы разделить выручку каждой когорты на её размер.
ltv = cumulative_rev.div(cohort_size, axis=0)

# Расчет ROI (Return on Investment)
roi = ltv.div(monthly_cac, axis=0) - 1

# 4. ФОРМИРОВАНИЕ ЛЕСЕНКИ ДЛЯ ВИЗУАЛИЗАЦИИ
# Для каждой когорты определяем максимальный прожитый месяц (когда был последний заказ)
max_n_by_cohort = cohort_revenue.apply(lambda row: row[row > 0].index.max() if row.sum() > 0 else 0, axis=1)

# Создаем копию таблицы ROI, которую будем превращать в лесенку.
roi_ladder = roi.copy()

# Проходим по каждой когорте и заменяем значения на NaN для месяцев, которые еще не наступили.
# Это делается для того, чтобы на тепловой карте не было призрачных данных о будущем.
for cohort in roi_ladder.index:
    max_n = max_n_by_cohort.get(cohort, 0) # Получаем последний прожитый месяц для когорты
    for n in roi_ladder.columns:
        if n > max_n: # Если номер месяца больше возраста когорты...
            roi_ladder.loc[cohort, n] = np.nan # ...заменяем значение на NaN.

# Приводим таблицу к стандартному виду: столбцы от 0 до 12 месяцев.
# reindex добавляет недостающие столбцы, fill_value=np.nan заполняет их отсутствующими данными (NaN).
# round(2) округляет значения до двух знаков после запятой для удобства чтения.
roi_ladder = roi_ladder.reindex(columns=range(0, 13), fill_value=np.nan).round(2)
```

### Каналы привлечения

Оцениваем эффективность каждого канала: сколько привлекли пользователей, во сколько это обошлось (CAC), какой ROI и конверсия. 

Цель – найти прибыльные каналы для масштабирования и убыточные – для принятия мер.

```python
# 1. Получаем каналы пользователей
user_channels = users[['user_id', 'acq_channel']].set_index('user_id')

# 2. Добавляем канал к заказам
orders_with_channel = orders[['user_id', 'revenue', 'order_id']].copy()
orders_with_channel['acq_channel'] = orders_with_channel['user_id'].map(user_channels['acq_channel'])

# 3. Агрегация по каналам (только по заказавшим)
channel_perf = (
    orders_with_channel
    .groupby('acq_channel')
    .agg(
        buyers=('order_id', 'nunique'),
        unique_buyers=('user_id', 'nunique'),
        revenue=('revenue', 'sum')
    )
    .join(
        users.groupby('acq_channel').size().rename('new_users'),
        how='right'
    )
    .join(
        campaigns.groupby('acq_channel')['budget'].sum(),
        how='left'
    )
    .fillna(0)
)

channel_perf['revenue'] = channel_perf['revenue'].round().astype(int)
channel_perf['budget'] = channel_perf['budget'].round().astype(int)

# 4. Расчёты
channel_perf['cr'] = channel_perf['unique_buyers'] / channel_perf['new_users']
channel_perf['cac'] = channel_perf['budget'] / channel_perf['new_users']
channel_perf['ltv_est'] = channel_perf['revenue'] / channel_perf['new_users']
channel_perf['roi'] = channel_perf['ltv_est'] / channel_perf['cac'] - 1

# Сортируем
channel_perf = channel_perf.sort_values('roi', ascending=False)

# Выводим
display(channel_perf)
```

```python
# ГРАФИК 1. ROI по каналам
ch = channel_perf.reset_index().sort_values('roi', ascending=True)

# Нормализация и цвета
norm = plt.Normalize(ch['roi'].min(), ch['roi'].max())
colors = plt.cm.RdYlGn(norm(ch['roi']))

# Построение графика
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(ch['acq_channel'], ch['roi'] * 100, color=colors)
ax.set_xlabel('ROI, %')
ax.set_title('ROI по каналам привлечения', fontweight='bold')
ax.axvline(0, color='black', linewidth=0.8, linestyle='-')

# Аннотации для столбцов
for bar, roi in zip(bars, ch['roi']):
    value_pct = roi * 100
    offset = 3 * (1 if value_pct >= 0 else -1)
    ha = 'left' if value_pct >= 0 else 'right'
    
    ax.text(
        bar.get_width() + offset,
        bar.get_y() + bar.get_height() / 2,
        f'{value_pct:+.0f}%',
        va='center',
        ha=ha,
        fontweight='bold',
        fontsize=10,
        color='black'
    )

plt.tight_layout()
plt.show()
```

```python
# Анализ эффективности по регионам
# 1. Собираем данные: пользователи + их заказы (выручка)
reg = (
    users[['user_id', 'region']]
    .merge(orders[['user_id', 'revenue', 'order_id']], on='user_id', how='left')
    .groupby('region')
    .agg(
        users=('user_id', 'nunique'),
        orders=('order_id', 'count'),
        revenue=('revenue', 'sum')
    )
)
# 2. Форматируем и считаем производные метрики
reg['revenue'] = reg['revenue'].round().astype(int)
reg['cr'] = reg['orders'] / reg['users']
reg['aov'] = reg['revenue'] / reg['orders']

# 3. Вывод таблицы
display(reg)
```

### Категории товаров

Анализируем ассортимент: какие категории генерируют выручку, какие – аутсайдеры. Помогает расставить приоритеты для закупок и промо.

```python
# Анализ категорий товаров
# Считаем ключевые метрики по каждой категории:
cat = (
    orders
    .groupby('category_name')
    .agg(
        orders=('order_id', 'nunique'),
        revenue=('revenue', 'sum'),
        aov=('total_price', 'mean'),
        users=('user_id', 'nunique')
    )
)

cat['revenue'] = cat['revenue'].round().astype(int)
cat['aov'] = cat['aov'].round(0).astype(int)
cat['revenue_share'] = (cat['revenue'] / cat['revenue'].sum()).round(4)

cat = cat.sort_values('revenue', ascending=False)
display(cat)
```

### Анализ покупательских сегментов (RFM)

Если регионы показывают **где** мы теряем выручку, то сегментация покупателей отвечает на вопрос **кто** эти пользователи и как они себя ведут.

Разбиваем всех пользователей на 4 сегмента по частоте покупок:
- **one_time** – совершили ровно 1 заказ
- **rare** – 2–3 заказа
- **medium** – 4–6 заказов
- **regular** – 7+ заказов

**Цель:** понять, какие сегменты приносят деньги, а какие требуют доработки, и куда направить усилия по удержанию.

```python
# Агрегация метрик по сегментам покупателей
# Джойним пользователей с заказами и группируем по сегменту
seg = (
    users[['user_id', 'buyer_segment']]
    .merge(orders[['user_id', 'revenue', 'order_id', 'order_date']], on='user_id', how='left')
    .groupby('buyer_segment')
    .agg(
        users=('user_id', 'nunique'),
        orders=('order_id', 'count'),
        revenue=('revenue', 'sum'),
        first_order=('order_date', 'min'),
        last_order=('order_date', 'max')
    )
)

# Приводим денежные метрики к целым и считаем производные показатели
seg['revenue'] = seg['revenue'].round().astype(int)
seg['orders_per_user'] = seg['orders'] / seg['users']
seg['revenue_per_user'] = (seg['revenue'] / seg['users']).round(0).astype(int)
# Выводим таблицу
display(seg)
```

**Выводы по сегментам покупателей:**

- **Regular** (7 773 пользователя) – основа бизнеса: 67% выручки, 2.7 заказа на пользователя.
- **Medium** (16 421) – второй по значимости, но частота покупок низкая (0.46).
- **One_time** (6 701, 15% аудитории) – проблема: лишь 1% выручки, 33 ₽ на пользователя.

### Воронка конверсии

Смотрим путь пользователя от первого визита до оформления заказа. Находим этапы с максимальной потерей – это зоны для A/B-тестов.

```python
# Возьмём пользователей, которые дошли до checkout_complete ИЛИ сделали заказ
buyers_ids = set(orders['user_id']) | set(events[events['event_type'] == 'checkout_complete']['user_id'])

funnel = (
    events[events['user_id'].isin(buyers_ids)]
    .groupby('event_type')['user_id'].nunique()
    .sort_values(ascending=False)
)

display(funnel)
```

```python
# АНАЛИЗ ВОРОНКИ КОНВЕРСИИ 

# 1. Определяем этапы воронки в правильном порядке
funnel_steps = [
    'page_view',        
    'product_view',     
    'add_to_cart',      
    'checkout_start',    
    'checkout_complete'  
]

# 2. Считаем уникальных пользователей на каждом этапе среди всех
funnel_all = {}
for step in funnel_steps:
    if step in events['event_type'].values:
        count = events[events['event_type'] == step]['user_id'].nunique()
    else:
        count = 0
    funnel_all[step] = count

# 3. Формируем DataFrame
funnel_df = pd.DataFrame({
    'step': funnel_steps,
    'users': [funnel_all[step] for step in funnel_steps]
})

# 4. Добавляем MAU (все, кто был в events за период)
mau_total = events['user_id'].nunique()
funnel_df = pd.concat([
    pd.DataFrame({'step': ['MAU'], 'users': [mau_total]}),
    funnel_df
], ignore_index=True)

# 5. Считаем конверсии
funnel_df['% от MAU'] = (funnel_df['users'] / funnel_df.loc[0, 'users'] * 100).round(1)
funnel_df['% от предыдущего'] = (
    funnel_df['users'] / funnel_df['users'].shift(1) * 100
).round(1)
funnel_df.loc[0, '% от предыдущего'] = 100.0

# 6. Подписи для визуализации
funnel_df['label_prev'] = funnel_df['% от предыдущего'].astype(str) + '%'
funnel_df['label_mau'] = funnel_df['% от MAU'].astype(str) + '%'
```

```python
# Интерактивная воронка с plotly
fig = go.Figure()

fig.add_trace(go.Funnel(
    name='Воронка конверсии',
    y=funnel_df['step'].tolist(),
    x=funnel_df['users'].tolist(),
    textposition="inside",
    textinfo="value+percent initial",
    texttemplate='<b>%{value:,.0f} чел.</b><br><i>%{percentInitial:.1%}</i>',
    marker=dict(
        color=['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728', '#9467bd', '#8c564b'],
        line=dict(width=2, color='white')
    ),
    connector=dict(line=dict(color='gray', width=1, dash='dot'))
))

# Добавляем аннотации с конверсией от предыдущего этапа
for i in range(len(funnel_df)-1):
    conv_rate = funnel_df.iloc[i+1]['% от предыдущего']
    
    # Определяем цвет в зависимости от величины падения
    if conv_rate < 80:
        color = 'red'
    elif conv_rate < 90:
        color = 'orange'
    else:
        color = 'green'
    
    fig.add_annotation(
        x=0.85,
        y=len(funnel_df) - i - 1.5,
        text=f'<b>{conv_rate:.1f}%</b>',
        showarrow=True,
        arrowhead=2,
        arrowcolor=color,
        arrowsize=1.5,
        arrowwidth=2,
        font=dict(size=13, color=color),
        bgcolor='white',
        bordercolor='lightgray',
        borderwidth=1,
        borderpad=4
    )

fig.update_layout(
    title={
        'text': '<b>Воронка конверсии</b><br><sup>Проценты справа — конверсия от предыдущего этапа</sup>',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 16, 'family': 'Arial'}
    },
    height=600,
    showlegend=False,
    annotations=[
        dict(
            x=0.5,
            y=-0.12,
            xref='paper',
            yref='paper',
            text='<b>⚠️ Критическое падение:</b> добавление в корзину → оформление заказа',
            showarrow=False,
            font=dict(size=12, color='red'),
            bgcolor='#ffe6e6',
            borderpad=10
        )
    ]
)

fig.show()
```

```python
# Динамика потерь и конверсий
# Подготавливаем данные для дополнительных графиков
stages = funnel_df['step'].tolist()
funnel_users = funnel_df['users'].tolist()  # переименовано, чтобы не конфликтовать с df users
drops = [funnel_users[i-1] - funnel_users[i] for i in range(1, len(funnel_users))]

# Каскадная диаграмма потерь и конверсия между этапами
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Левая часть: каскад потерь
x_pos = np.arange(len(stages))

# Цвета для потерь
drop_colors = []
for i in range(len(drops)):
    pct = (drops[i] / funnel_users[i]) * 100
    if pct < 5:
        drop_colors.append('#2ecc71')
    elif pct < 10:
        drop_colors.append('#f39c12')
    else:
        drop_colors.append('#e74c3c')

# Столбцы с пользователями
ax1.bar(x_pos, funnel_users, color='#3498db', alpha=0.8, zorder=2, label='Пользователи')

# Столбцы с потерями
for i in range(len(drops)):
    ax1.bar(x_pos[i+1], drops[i], bottom=funnel_users[i+1], color=drop_colors[i], 
            alpha=0.6, zorder=2, label='Потери' if i == 0 else '')

# Подписи на барах
for i, (stage, user) in enumerate(zip(stages, funnel_users)):
    ax1.text(i, user/2, f'{user:,}', ha='center', va='center', 
            fontweight='bold', color='white', fontsize=10)

# Подписи потерь
for i, drop in enumerate(drops):
    if drop > 0:
        pct = (drop / funnel_users[i]) * 100
        ax1.text(i+1, funnel_users[i+1] + drop/2, f'-{pct:.1f}%\n({drop:,})', 
                ha='center', va='center', fontweight='bold', color='darkred', fontsize=8)

ax1.set_xticks(x_pos)
ax1.set_xticklabels(stages, rotation=45, ha='right')
ax1.set_ylabel('Количество пользователей')
ax1.set_title('Каскадная диаграмма потерь', fontweight='bold', fontsize=13)
ax1.grid(axis='y', alpha=0.3)
ax1.set_ylim(0, max(funnel_users) * 1.3)
ax1.legend(loc='upper right')

# Конверсия между этапами
conv_rates = funnel_df['% от предыдущего'].tolist()[1:]  # без первого (100%)
conv_stages = [f'{stages[i]} → {stages[i+1]}' for i in range(len(stages)-1)]

colors_conv = ['#e74c3c' if x < 80 else '#f39c12' if x < 90 else '#2ecc71' for x in conv_rates]
bars = ax2.barh(conv_stages, conv_rates, color=colors_conv, alpha=0.8)

for bar, rate in zip(bars, conv_rates):
    ax2.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2, 
            f'{rate:.1f}%', va='center', fontweight='bold', fontsize=11)

ax2.axvline(x=80, color='red', linestyle='--', alpha=0.5, linewidth=1, label='Критический порог (80%)')
ax2.axvline(x=90, color='orange', linestyle='--', alpha=0.5, linewidth=1, label='Предупреждение (90%)')
ax2.set_xlabel('Конверсия (%)')
ax2.set_title('Конверсия между этапами', fontweight='bold', fontsize=13)
ax2.set_xlim(0, 105)
ax2.legend(loc='lower right')

plt.tight_layout()
plt.show()
```

**Вывод по общей воронке конверсии:**

- **До добавления в корзину** – потерь почти нет (99.7% доходят до `add_to_cart`).
- **Критический этап 1:** `add_to_cart → checkout_start` – теряем 7.5 п.п. (92.5% → 82.2%).
- **Критический этап 2:** `checkout_start → checkout_complete` – теряем ещё 17.1 п.п. (82.2% → 82.9% конверсии от предыдущего, но в абсолюте – 2 872 пользователя).

**Итог:** общая конверсия от MAU до покупки – 62.8%. Основные потери на этапах оформления заказа (корзина → начало оформления → завершение). Это зоны для A/B-тестов.

```python
# ВОРОНКА ПО СЕГМЕНТАМ
# ПОДГОТОВКА ДАННЫХ 
funnel_steps = ['page_view', 'product_view', 'add_to_cart', 'checkout_start', 'checkout_complete']

# Привязываем месяц и сегмент к событиям
events['event_month'] = pd.to_datetime(events['event_date']).dt.to_period('M').astype(str)
users_seg = users[['user_id', 'buyer_segment']]
events_with_seg = events.merge(users_seg, on='user_id', how='left')

# Считаем среднюю конверсию по каждому сегменту и этапу
segment_funnel = []

for segment in events_with_seg['buyer_segment'].unique():
    if pd.isna(segment):
        continue
    
    sub_seg = events_with_seg[events_with_seg['buyer_segment'] == segment]
    mau_by_month = sub_seg.groupby('event_month')['user_id'].nunique()
    avg_mau = mau_by_month.mean()
    
    for step in funnel_steps:
        users_by_month = sub_seg[sub_seg['event_type'] == step].groupby('event_month')['user_id'].nunique()
        avg_users = users_by_month.mean()
        
        segment_funnel.append({
            'segment': segment,
            'stage': step,
            'avg_users': avg_users,
            'avg_mau': avg_mau
        })

funnel_agg = pd.DataFrame(segment_funnel)
funnel_agg = funnel_agg.dropna(subset=['segment'])

# Конверсия от MAU
funnel_agg['conv_from_mau'] = (funnel_agg['avg_users'] / funnel_agg['avg_mau'] * 100).round(1)
```

```python
# Создание фигуры и добавление воронок для каждого сегмента
fig = go.Figure()

# Цвета для сегментов
segment_colors = {
    'one_time': '#FF6B6B',
    'rare': '#4ECDC4', 
    'medium': '#45B7D1',
    'regular': '#96CEB4'
}

# Этапы воронки в правильном порядке
stages = ['page_view', 'product_view', 'add_to_cart', 'checkout_start', 'checkout_complete']

for segment in funnel_agg['segment'].unique():
    # Фильтруем данные для одного сегмента
    segment_data = funnel_agg[funnel_agg['segment'] == segment]
    # Извлекаем значения конверсии от MAU для каждого этапа
    values = []
    for stage in stages:
        val = segment_data[segment_data['stage'] == stage]['conv_from_mau'].values
        values.append(val[0] if len(val) > 0 else 0)
    
    # Добавляем слой-воронку для сегмента
    fig.add_trace(go.Funnel(
        name=segment,
        y=stages,
        x=values,
        textinfo="value+percent initial",
        textposition="inside",
        marker={"color": segment_colors.get(segment, 'grey')},
        opacity=0.85
    ))

# Настройка оформления
fig.update_layout(
    title={
        'text': 'Сравнение воронок конверсии по сегментам покупателей (% от MAU)',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 16, 'weight': 'bold'}
    },
    funnelgap=0.3, 
    height=600,
    width=1100,
    legend_title_text='Сегмент',
    legend=dict(
        yanchor="middle",
        y=0.5,
        xanchor="left",
        x=1.02,
        bgcolor='rgba(255,255,255,0.8)'
    ),
    margin=dict(r=150)  # добавляем отступ справа для легенды
)

fig.show()
```

**ВЫВОДЫ ПО СЕГМЕНТАМ**

1. **Сегмент 'regular' лидирует по конверсии** на всех этапах (в 2-3 раза выше `one_time`)

2. **Критическая точка оттока для всех сегментов:** `add_to_cart → checkout_start`
   - У `regular` теряем **60%** пользователей
   - У `one_time` теряем **85-90%**

3. **`one_time` практически не доходит до оформления заказа** - проблема либо в качестве привлечения, либо в UX для новых пользователей

4. **Рекомендуемые гипотезы:**
   - Упростить оформление заказа (убрать лишние поля, добавить гостевой checkout)
   - Внедрить exit-intent popup с промокодом при уходе с корзины
   - Для `one_time` - тестировать разные онбординги и первый заказ со скидкой

```python
# АНАЛИЗ ПО СЕГМЕНТАМ ПОКУПАТЕЛЕЙ

if 'revenue' not in orders.columns:
    orders['revenue'] = orders['total_price'] * 0.05

users_seg = users[['user_id', 'buyer_segment', 'cohort_month']].copy()
orders_enh = orders[['user_id', 'revenue']].copy()
orders_enh['order_month'] = pd.to_datetime(orders['order_date']).dt.to_period('M').astype(str)

monthly_seg = (
    orders_enh.merge(users_seg[['user_id', 'buyer_segment']], on='user_id', how='inner')
    .groupby(['order_month', 'buyer_segment'])
    .agg(revenue=('revenue', 'sum'))
    .reset_index()
)

users_monthly = (
    users_seg.groupby(['cohort_month', 'buyer_segment'])
    .size()
    .rename('users')
    .reset_index()
)
users_monthly['cohort_month'] = pd.to_datetime(users_monthly['cohort_month']).dt.to_period('M').astype(str)

months = sorted(set(monthly_seg['order_month'].unique()) | set(users_monthly['cohort_month'].unique()))
segments = ['one_time', 'rare', 'medium', 'regular']

monthly_pivot = (
    monthly_seg.set_index(['order_month', 'buyer_segment'])
    .reindex(pd.MultiIndex.from_product([months, segments], names=['month', 'buyer_segment']), fill_value=0)
    .reset_index()
)
users_pivot = (
    users_monthly.set_index(['cohort_month', 'buyer_segment'])
    .reindex(pd.MultiIndex.from_product([months, segments], names=['month', 'buyer_segment']), fill_value=0)
    .reset_index()
)

monthly_pivot = monthly_pivot.rename(columns={'month': 'month'})
users_pivot = users_pivot.rename(columns={'month': 'month'})

full_pivot = monthly_pivot.merge(users_pivot, on=['month', 'buyer_segment'], suffixes=('', '_users'))
full_pivot = full_pivot.rename(columns={'users_users': 'users'})

# Визуализация

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10))

colors = {
    'one_time': '#FF6B6B',
    'rare': '#4ECDC4',
    'medium': '#45B7D1',
    'regular': '#96CEB4'
}

# 1. ГРАФИК ПОЛЬЗОВАТЕЛЕЙ
for segment in segments:
    data = full_pivot[full_pivot['buyer_segment'] == segment].sort_values('month')
    ax1.plot(data['month'], data['users'].values, 
             marker='o', linewidth=2, markersize=6,
             label=segment, color=colors[segment])

ax1.set_xlabel('Месяц')
ax1.set_ylabel('Количество пользователей')
ax1.set_title('Динамика пользователей по сегментам (по месяцу регистрации)', fontweight='bold')
ax1.legend(loc='center left', bbox_to_anchor=(1, 0.5), title='Сегмент')
ax1.grid(True, alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# 2. ГРАФИК ВЫРУЧКИ
for segment in segments:
    data = full_pivot[full_pivot['buyer_segment'] == segment].sort_values('month')
    ax2.plot(data['month'], data['revenue'].values, 
             marker='s', linewidth=2, markersize=6,
             label=segment, color=colors[segment])

ax2.set_xlabel('Месяц')
ax2.set_ylabel('Выручка')
ax2.set_title('Динамика выручки по сегментам (по месяцу заказа)', fontweight='bold')
ax2.legend(loc='center left', bbox_to_anchor=(1, 0.5), title='Сегмент')
ax2.grid(True, alpha=0.3)
ax2.tick_params(axis='x', rotation=45)

# Отступ справа для легенды
plt.subplots_adjust(right=0.85)
plt.tight_layout()
plt.show()
```

Данные готовы к обработке, пропусков нет, типы данных верны, группы для A/B теста равны.

```python
users_with_group = users_ab.merge(ab_split, on='user_id', how='inner')

# Фильтруем только TikTok
users_tiktok = users_with_group[users_with_group['acq_channel'] == 'TikTok']
print(f"Пользователей TikTok в сплите: {len(users_tiktok)}")

# Добавляем заказы
orders_tiktok = orders_ab[orders_ab['user_id'].isin(users_tiktok['user_id'])]
print(f"Заказов от TikTok: {len(orders_tiktok)}")

# Считаем метрики на пользователя
user_metrics = (
    users_tiktok[['user_id', 'group']]
    .merge(
        orders_tiktok.groupby('user_id').agg(
            first_order=('order_date', 'min'),
            total_revenue=('total_price', 'sum'),
            orders=('order_id', 'nunique')
        ).reset_index(),
        on='user_id',
        how='left'
    )
    .fillna({'first_order': pd.NaT, 'total_revenue': 0, 'orders': 0})
)

# Конверсия в первую покупку
user_metrics['has_first_order'] = (user_metrics['orders'] >= 1).astype(int)
```

**Анализ гипотез A/B-теста**

** 🎯 Основная метрика: конверсия в первую покупку**

**H₀ (нулевая гипотеза)**  
Предоставление бонуса на первую покупку пользователям, привлечённым через TikTok, **не оказывает статистически значимого влияния** на конверсию.  
`CR_control ≥ CR_treatment`

**H₁ (альтернативная гипотеза)**  
Бонус **увеличивает конверсию** в первую покупку.  
`CR_treatment > CR_control`

---

**Барьерная метрика: общая выручка (`total_price`)**

**H₀**  
Выручка в тестовой группе **не выше**, чем в контрольной.  
`total_revenue_treatment ≤ total_revenue_control`

**H₁**  
Выручка в тестовой группе **выше**.  
`total_revenue_treatment > total_revenue_control`

---

**Вспомогательные метрики**

**1. ARPPU (выручка на покупателя)**

**H₀**  
`ARPPU_treatment ≤ ARPPU_control`

**H₁**  
`ARPPU_treatment > ARPPU_control`

---

**2. AOV (средний чек)**

**H₀**  
`AOV_treatment ≤ AOV_control`

**H₁**  
`AOV_treatment > AOV_control`

```python
group_stats = (
    user_metrics
    .groupby('group', as_index=False)
    .agg(
        users=('user_id', 'count'),
        buyers=('has_first_order', 'sum'),
        total_revenue=('total_revenue', 'sum'),  # total_price — полная сумма заказов
        orders=('orders', 'sum')
    )
)

# Расчёт метрик
group_stats['cr_first_order'] = group_stats['buyers'] / group_stats['users']
group_stats['arpu'] = group_stats['total_revenue'] / group_stats['users']
group_stats['arppu'] = group_stats['total_revenue'] / group_stats['buyers'].replace(0, np.nan)
group_stats['aov'] = group_stats['total_revenue'] / group_stats['orders'].replace(0, np.nan)

# Вывод таблицы (округление только для отображения)
display(group_stats.round({
    'cr_first_order': 4,
    'arpu': 2,
    'arppu': 2,
    'aov': 2
}))
```

Бонус увеличил конверсию с 1% до 7,5% и поднял выручку на пользователя в 10 раз, при этом средний чек практически не изменился – значит, пользователи не просто дробили покупки под бонус, а реально начали покупать.

```python
A = group_stats[group_stats['group'] == 'control'].iloc[0]
B = group_stats[group_stats['group'] == 'treatment'].iloc[0]

p1 = A['buyers'] / A['users']   # control
p2 = B['buyers'] / B['users']   # treatment
n1 = A['users']
n2 = B['users']

# Объединённая пропорция (при H0)
p_pool = (A['buyers'] + B['buyers']) / (n1 + n2)

# Стандартная ошибка
se = np.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

# z-статистика (односторонняя: B > A)
z = (p2 - p1) / se
p_value = stats.norm.sf(z)  # односторонний

print(f"z = {z:.3f}, p-value = {p_value:.4e}")

print(f"Конверсия control: {A['cr_first_order']:.2%}")
print(f"Конверсия treatment: {B['cr_first_order']:.2%}")
print(f"Рост: +{((B['cr_first_order'] / A['cr_first_order']) - 1):.1%}")
print(f"p-value: {p_value:.4e}")
print(f"Значимо при α=0.05? {'Да' if p_value < 0.05 else 'Нет'}")
```

Бонус статистически значимо (p < 0.0001) увеличил конверсию с 1,06% до 7,51% — рост в 7 раз. Нулевая гипотеза отвергается, эффект доказан.

```python
# СРАВНЕНИЕ ВЫРУЧКИ НА ПОЛЬЗОВАТЕЛЯ (ARPU) МЕЖДУ ГРУППАМИ

# Извлекаем выручку каждого пользователя отдельно для контрольной и тестовой групп
revenue_control = user_metrics[user_metrics['group'] == 'control']['total_revenue']
revenue_treatment = user_metrics[user_metrics['group'] == 'treatment']['total_revenue']

# Преобразуем в массивы для удобства вычислений
x1 = revenue_control.values  # контрольная группа
x2 = revenue_treatment.values  # тестовая группа

# Базовые статистики
n1, n2 = len(x1), len(x2)          # размеры групп
mean1, mean2 = np.mean(x1), np.mean(x2)  # средняя выручка на пользователя
var1, var2 = np.var(x1, ddof=1), np.var(x2, ddof=1)  # дисперсии (ddof=1 — несмещённая)

# t-статистика Уэлча (не предполагает равенства дисперсий)
# Формула: разница средних / стандартная ошибка разницы
t = (mean2 - mean1) / np.sqrt(var1/n1 + var2/n2)

# Степени свободы по Уэлчу-Саттертуэйту (корректировка для неравных дисперсий)
df = (var1/n1 + var2/n2)**2 / ((var1/n1)**2/(n1-1) + (var2/n2)**2/(n2-1))

# Односторонний p-value: проверяем гипотезу, что mean2 > mean1
# (бонус увеличил выручку на пользователя)
p_value = 1 - stats.t.cdf(t, df)

# Вывод результатов
print(f"t = {t:.3f}")           # t-статистика: чем больше, тем сильнее разница
print(f"df = {df:.1f}")         # степени свободы (нужны для интерпретации)
print(f"p-value = {p_value:.2e}")  # вероятность ошибки при отвержении H0
```

Разница в ARPU между группами статистически значима (p < 0.001). Нулевая гипотеза о равенстве средних отвергается.

```python
# СРАВНЕНИЕ ARPPU (ВЫРУЧКА НА ПОКУПАТЕЛЯ) МЕЖДУ ГРУППАМИ
# Берём только покупателей (has_first_order == 1) в каждой группе
revenue_control_buyers = user_metrics[
    (user_metrics['group'] == 'control') & (user_metrics['has_first_order'] == 1)
]['total_revenue'].values

revenue_treatment_buyers = user_metrics[
    (user_metrics['group'] == 'treatment') & (user_metrics['has_first_order'] == 1)
]['total_revenue'].values

# Базовые статистики по покупателям
n_ctrl, n_trt = len(revenue_control_buyers), len(revenue_treatment_buyers)
mean_ctrl, mean_trt = revenue_control_buyers.mean(), revenue_treatment_buyers.mean()
var_ctrl, var_trt = revenue_control_buyers.var(ddof=1), revenue_treatment_buyers.var(ddof=1)

arppu_ctrl, arppu_trt = mean_ctrl, mean_trt
print(f"ARPPU control  (n={n_ctrl}):  {arppu_ctrl:,.2f}")
print(f"ARPPU treatment (n={n_trt}): {arppu_trt:,.2f}")
print(f"Рост: +{(arppu_trt / arppu_ctrl - 1):.1%}")

# Welch’s t-test (для неравных дисперсий)
t_stat = (mean_trt - mean_ctrl) / np.sqrt(var_trt / n_trt + var_ctrl / n_ctrl)

# Степени свободы по Уэлчу–Саттертуэйту
df_num = (var_trt / n_trt + var_ctrl / n_ctrl) ** 2
df_den = (var_trt / n_trt) ** 2 / (n_trt - 1) + (var_ctrl / n_ctrl) ** 2 / (n_ctrl - 1)
df = df_num / df_den

# Односторонний p-value: проверяем, что treatment > control
p_value = 1 - stats.t.cdf(t_stat, df)

print(f"\nWelch’s t-test (вручную):")
print(f"t = {t_stat:.3f}, df = {df:.1f}, p-value = {p_value:.2e}")

# Вывод
alpha = 0.05
if p_value < alpha:
    print("H0 отвергается: ARPPU в тестовой группе значимо выше.")
else:
    print("Недостаточно доказательств, чтобы отвергнуть H0.")
```

Бонус статистически значимо повысил ARPPU на 47.8% (с 23 648 ₽ до 34 960 ₽). Покупатели из тестовой группы тратят в среднем больше, чем из контрольной (p = 0.0246). Это опровергает риск того, что бонус привлекает дешёвых покупателей, наоборот, качество платящей аудитории улучшилось.

```python
# СРАВНЕНИЕ AOV (СРЕДНЕГО ЧЕКА) МЕЖДУ ГРУППАМИ
# Отбираем только TikTok-пользователей и их заказы
tiktok_users = set(users_ab[users_ab['acq_channel'] == 'TikTok']['user_id'])
orders_tiktok = orders_ab[orders_ab['user_id'].isin(tiktok_users)]

# Добавляем метку группы (control/treatment) к каждому заказу
orders_with_group = orders_tiktok.merge(
    ab_split[['user_id', 'group']],
    on='user_id',
    how='inner'
)

# Извлекаем суммы заказов по группам
aov_control = orders_with_group[orders_with_group['group'] == 'control']['total_price'].values
aov_treatment = orders_with_group[orders_with_group['group'] == 'treatment']['total_price'].values

# Средние значения AOV
aov_ctrl_mean = aov_control.mean()
aov_trt_mean = aov_treatment.mean()

n_ctrl, n_trt = len(aov_control), len(aov_treatment)

print(f"AOV control  (n={n_ctrl}):  {aov_ctrl_mean:,.2f}")
print(f"AOV treatment (n={n_trt}): {aov_trt_mean:,.2f}")
print(f"Рост: +{(aov_trt_mean / aov_ctrl_mean - 1):.1%}")

# T-test Уэлча (односторонний: проверяем, что средний чек не упал)
# H0: AOV_treatment ≤ AOV_control (бонус снизил или не изменил чек)
# H1: AOV_treatment > AOV_control (бонус не навредил чеку)
mean_ctrl, mean_trt = aov_ctrl_mean, aov_trt_mean
var_ctrl, var_trt = aov_control.var(ddof=1), aov_treatment.var(ddof=1)

t_stat = (mean_trt - mean_ctrl) / np.sqrt(var_trt / n_trt + var_ctrl / n_ctrl)

df_num = (var_trt / n_trt + var_ctrl / n_ctrl) ** 2
df_den = (var_trt / n_trt) ** 2 / (n_trt - 1) + (var_ctrl / n_ctrl) ** 2 / (n_ctrl - 1)
df = df_num / df_den

p_value = 1 - stats.t.cdf(t_stat, df)

print(f"\nWelch’s t-test (AOV, по заказам):")
print(f"t = {t_stat:.3f}, df = {df:.1f}, p-value = {p_value:.2e}")

# Вывод
alpha = 0.05
if p_value < alpha:
    print("H0 отвергается: AOV в тестовой группе значимо выше.")
else:
    print("Недостаточно доказательств для отклонения H0 (AOV значимо не изменился).")
```

### 1. Ключевые результаты эксперимента

Тест проводился на пользователях, привлечённых через TikTok. Целевая группа получала бонус на первую покупку, контрольная – нет. Получены следующие результаты:

| Метрика | Контрольная группа | Тестовая группа | Изменение |
|---------|--------------------|-----------------|-----------|
| **CR (конверсия в первую покупку)** | 1.06% | 7.51% | **▲ +609%** |
| **ARPU** | 252 ₽ | 2 625 ₽ | **▲ ×10.4** |
| **AOV (средний чек)** | 15 766 ₽ | 16 648 ₽ | **стабилен (p=0.394)** |
| **ARPPU** | 23 648 ₽ | 34 960 ₽ | **▲ +48% (p=0.025)** |

**Статистическая значимость:** p-value < 0.0001 – нулевая гипотеза (отсутствие эффекта) отвергается.

### 2. Power Analysis (чувствительность теста)

Оценка минимального детектируемого эффекта (MDE) при текущем размере выборки:

| Параметр | Значение |
|----------|----------|
| Размер выборки | 1 465 пользователей в группе |
| Базовая конверсия | 1.06% |
| Уровень значимости (α) | 0.05 |
| Статистическая мощность (1-β) | 80% |
| **MDE** | **+9.19 п.п.** |
| Фактический эффект | +6.45 п.п. (1.06% → 7.51%) |

**Интерпретация:**

Фактический эффект оказался ниже расчётного MDE – технически тест был недостаточно мощный для гарантированного обнаружения эффекта данного размера. Однако:

- Эффект **кратный** (рост в 7 раз)
- p-value < 0.0001

Этого достаточно, чтобы считать результат надёжным. При масштабировании рекомендуется подтвердить на большей выборке (n ≥ 2 000).

### 3. Интерпретация и инсайты

1. **Бонус радикально повышает конверсию** – рост в 7 раз свидетельствует о высокой чувствительности аудитории TikTok к ценовым стимулам на старте.
2. **Рост ARPU кратен росту конверсии** – это подтверждает, что привлечённые пользователи не просто совершают мелкие пробные покупки, а демонстрируют полноценное покупательское поведение.
3. **AOV остался неизменным** – ключевой позитивный сигнал. Рост конверсии и выручки достигнут без ценового демпинга, за счёт вовлечения ранее не конвертировавших пользователей.
4. **ARPPU вырос на 48%** – покупатели из тестовой группы тратят значимо больше, чем из контрольной. Бонус привлёк не просто больше покупателей, но и более качественных.

### 4. Рекомендации по внедрению

| № | Рекомендация | Обоснование |
|---|--------------|-------------|
| **1** | **Запустить бонус в продакшен для всего TikTok-трафика** | Экономическая эффективность доказана: выручка выросла в 10 раз при стабильном чеке. |
| **2** | **Обкатать бонус на других каналах с низкой конверсией** | Механизм может быть эффективен для аудиторий, схожих с TikTok по вовлечённости и порогу входа. |
| **3** | **Оптимизировать размер бонуса (A/B-тест нескольких значений)** | Текущий результат – «точка входа». Возможно, меньший бонус даст сопоставимый эффект с лучшей экономикой. |
| **4** | **Оценить LTV и удержание когорт с бонусом** | Необходимо убедиться, что пользователи, пришедшие по бонусу, не «отваливаются» после первой покупки. |
| **5** | **Интегрировать бонус в сценарии ретаргетинга** | Возможно, бонус эффективен и для «отказников» из других каналов – стоит протестировать. |

### 5. Ограничения эксперимента

1. **Горизонт анализа** – тест длился 3 недели. Часть пользователей из поздних когорт могла не успеть совершить покупку, что потенциально занижает конверсию в обеих группах.
2. **Затраты на бонус** – не вычтены из выручки. Финальный ROMI требует отдельного расчёта с учётом стоимости бонуса.
3. **Размер выборки покупателей** – 16 и 110 наблюдений. Несмотря на статистическую значимость, результаты желательно подтвердить на больших данных после масштабирования.
4. **Эксперимент не решает общий спад конверсии** – проблема падения CR с 43% до 24% (выявленная в разделе 2) требует отдельного исследования и не связана с каналом TikTok.
5. **Мощность теста** – при текущей выборке MDE составил 9.19 п.п., что выше фактического эффекта (6.45 п.п.). Тест был недостаточно мощный, но значимый p-value позволяет считать результат надёжным.

### 6. Заключение

Эксперимент **статистически значимо подтвердил гипотезу** о положительном влиянии бонуса на первую покупку для пользователей TikTok. Несмотря на то, что фактический эффект (+6.45 п.п.) оказался ниже расчётного MDE (+9.19 п.п.) при текущей мощности теста, кратный рост конверсии (в 7 раз) и p-value < 0.0001 делают результат убедительным.

Рекомендуется **немедленное масштабирование** механики на весь TikTok-трафик с параллельным тестированием её эффективности на других каналах привлечения, а также повторное подтверждение на большей выборке для снятия технических ограничений по мощности.
